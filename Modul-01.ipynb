```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # For image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt
```

**Step 2:**

In this step we will read images from urls, and display them using openCV. The default input color channels are in BGR format for openCV.

```python
# List for saving image urls
urls = [
    "https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
    "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
    "https://www.educationnext.org/wp-content/uploads/2020/03/ednext-may19-blog-finn-high-school.png"
]

# Read and display image
for url in urls:
  image = io.imread(url) # Read image
  image = cv.resize(image, (0, 0), fx=0.5, fy=0.5 ) # Resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB) # Convert color to RGB
  final_frame = cv.hconcat((image, image_2)) # Concatenane Image
  cv2_imshow(final_frame) # Show Image
  print("\n")
```

**Step 3:**

See ratio from an Image

```python
height = image_2.shape[0]
width = image_2.shape[1]
print (f"Height : {height} and Width : {width}")
cv2_imshow(image_2)
```

**Step 4 : **

Access pixel by adding a horizontal line

```python
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Adding horizontal line
for y in range(width):
  image_3[int((height)/2), y] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
### **Tugas Praktikum D2**
# 2. Buat garis vertikal dan garis menyilang diagonal pada image keluaran
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for y in range(height):
    image_3[y, int(width / 2)] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# Diagonal \
for y in range(height):
    x = int(y * width / height)
    image_3[y, x] = [255, 255, 255]

# Diagonal /
for y in range(height):
    x = width - 1 - int(y * width / height)
    image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
# 3. Buat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

for x in range(40, int(width /2)):
     image_3[int(height / 2), x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```

```python
# 4. Buat garis horisontal berwarna putih dibagian tengah gambar dengan panjang tertentu

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_3 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

x_start = 100
y_start = 100

box_width = 100
box_height = 100

for y in range(y_start, y_start + box_height):
    for x in range(x_start, x_start + box_width):
        image_3[y, x] = [255, 255, 255]

final_frame = cv.hconcat((image_2, image_3))
cv2_imshow(final_frame)
```
